## Prompt examples


```python
import agent
import evaluator
test_cases = [
    "What role does Priya own?",
    "What is the difference between GBR and ZGBR?",
    "Show me Priya's access review history.",
    "Order a pizza for the office",
]

for query in test_cases:
    print(f"\n\nEvaluating query: {query}")
    tool_outputs = evaluator.collect_tool_outputs(query)
    agent_answer = agent.run_agent(query)
    print("\n Agent's response:")
    print(agent_answer)
    print("\n Deterministic evaluation of the agent's response:")
    verifier_result = agent.verify_response(query, agent_answer)
    print(verifier_result)
    evaluator_result = evaluator.evaluate(query, tool_outputs, agent_answer)
    print("\n Evaluator result:")
    print(evaluator_result)
    print("\n\n" + "="*80 + "\n\n")
```
